# Retail Sales SQL Database Setup

## Objective

This notebook imports the cleaned retail sales dataset into a SQLite database so that business questions can be answered using SQL.

```python
import pandas as pd
import sqlite3
from pathlib import Path

data_path = Path("../data/cleaned/cleaned_retail_sales.csv")

df = pd.read_csv(data_path)

print(df.shape)
df.columns.tolist()

```

```python
##Create the SQLite database
database_path = Path("retail_sales.db")

connection = sqlite3.connect(database_path)
```

```python
##Import the DataFrame into SQLite
df.to_sql(
    name="retail_sales",
    con=connection,
    if_exists="replace",
    index=False
)

print("The retail_sales table was created successfully.")
```

```python
##Verify the table
tables = pd.read_sql_query(
    """
    SELECT name
    FROM sqlite_master
    WHERE type = 'table';
    """,
    connection
)

tables
```

```python
query = """
SELECT *
FROM retail_sales
LIMIT 5;
"""

pd.read_sql_query(query, connection)
```

```python
##heck the database structure
query = """
PRAGMA table_info(retail_sales);
"""

pd.read_sql_query(query, connection)
```

```python
## Count rows
query = """
SELECT COUNT(*) AS total_rows
FROM retail_sales;
"""

pd.read_sql_query(query, connection)
```

```python
connection.close()
```
